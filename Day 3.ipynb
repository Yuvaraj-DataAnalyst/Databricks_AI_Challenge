```python
#load the nov data and display 5 reocrds
df_sales_nov = spark.read.csv("/Volumes/workspace/ecommerce/ecommerce_data/2019-Nov.csv")
df_sales_nov = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/workspace/ecommerce/ecommerce_data/2019-Nov.csv")
df_sales_nov.limit(5).display()
```

# Sample Window Function

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col, sum

window_spec = Window.partitionBy("user_id").orderBy("event_time").rowsBetween(Window.unboundedPreceding, Window.currentRow)

```

```python
df_running_sum1 = df_sales_oct.withColumn(
    "running_sum_price",
    sum(col("price")).over(window_spec)
)
```

```python
display(df_running_sum1)


```

# Running total

```python
from pyspark.sql.functions import col, sum
from pyspark.sql.window import Window

df_inner1 = (
    df_sales_oct.alias("oct")
    .join(
        df_sales_nov.alias("nov"),
        ["product_id", "user_id"],
        "inner"
    )
)

window_spec = (
    Window
    .partitionBy("user_id")
    .orderBy(col("oct.event_time"))
    .rowsBetween(Window.unboundedPreceding, Window.currentRow)
)

df_running_total = df_inner1.withColumn(
    "running_total_price",
    sum(col("oct.price")).over(window_spec)
)

display(df_running_total)
```
