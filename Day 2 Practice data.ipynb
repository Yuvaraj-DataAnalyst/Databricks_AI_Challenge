```python
#LOAD DATA INTO DATAFRAME

df_sales_nov = spark.read.csv("/Volumes/workspace/ecommerce/ecommerce_data/2019-Nov.csv")
df_sales_nov.describe()
df_sales_nov.printSchema()
df_sales_nov.show()
```

```python
#ADDING EXISTING HEADER TO THE DATA
df_sales_nov = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/workspace/ecommerce/ecommerce_data/2019-Nov.csv")
```

```python
#SHOWING THE DATA
df_sales_nov.limit(10).show()
```

```python
#SELECTING THE COLUMN
df_sales_nov.select("event_time", "event_type", "price").show()
```

```python
#FILTER BY CONDITION
df_sales_nov.filter("event_type = 'purchase'").show()



```

```python
#GROUP BY CONDITION
df_sales_nov.groupBy("event_type").count().show()
```

```python
#ORDER BY CONDITION
df_sales_nov.orderBy("price", ascending=False).show(10)
```
